```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ✅ Create dummy dataset (like mining sensors: temperature, vibration, gas levels)
data = {
    "time": pd.date_range(start="2025-01-01", periods=100, freq="h"),
    "temperature": np.random.normal(30, 5, 100),  # Celsius
    "vibration": np.random.normal(0.5, 0.1, 100),  # g-force
    "gas_level": np.random.normal(400, 50, 100)   # ppm
}

df = pd.DataFrame(data)

# Show first 5 rows
print("Sample data:")
print(df.head())

# Plot
df.set_index("time").plot(subplots=True, figsize=(10,6), title="Dummy Mining Sensor Data")
plt.show()
```
